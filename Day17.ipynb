```python
# Import pandas for data manipulation
import pandas as pd

# Import TransactionEncoder to convert transactions into a binary matrix
from mlxtend.preprocessing import TransactionEncoder

# Import Apriori algorithm
from mlxtend.frequent_patterns import apriori

# Import association_rules function
from mlxtend.frequent_patterns import association_rules
```

```python
transactions = [
    ['Bread', 'Milk'],
    ['Bread', 'Diaper', 'Biscuits', 'Eggs'],
    ['Milk', 'Diaper', 'Biscuits', 'Coke'],
    ['Bread', 'Milk', 'Diaper', 'Biscuits'],
    ['Bread', 'Milk', 'Diaper', 'Coke']
]
```

```python
print(transactions)
```

te = TransactionEncoder()

te_array = te.fit(transactions).transform(transactions)

df = pd.DataFrame(te_array, columns=te.columns_)

```python
print(df)
```

```python
frequent_itemsets = apriori(
    df,
    min_support=0.4,
    use_colnames=True
)
```

```python
print(frequent_itemsets)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.7
)
```

```python
print(rules)
```

```python
print(
    rules[
        [
            'antecedents',
            'consequents',
            'support',
            'confidence',
            'lift'
        ]
    ]
)
```

```python
strong_rules = rules[
    rules['lift'] > 1
]

# print(strong_rules)

print(
   strong_rules[
        [
            'antecedents',
            'consequents',
            'support',
            'confidence',
            'lift'
        ]
    ]
)
```

```python
strong_rules = rules[rules['lift'] > 1]

# Create a new DataFrame with Rule and Lift
result = strong_rules[['antecedents', 'consequents', 'lift']].copy()

# Convert frozensets to readable text
result['Rule'] = (
    result['antecedents'].apply(lambda x: ', '.join(list(x))) +
    " →" +
    result['consequents'].apply(lambda x: ', '.join(list(x)))
)

# Keep only Rule and Lift columns
result = result[['Rule', 'lift']]

# Rename the lift column
result.rename(columns={'lift': 'Lift'}, inplace=True)

# Display the result
print(result)
```
