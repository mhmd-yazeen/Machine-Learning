### STEP 1 — IMPORT LIBRARIES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("lr_data1.csv")
```

### STEP 5 — DATA VISUALIZATION

```python
plt.scatter(df['YearsExperience'], df['Salary'],color="Green")
plt.title("Years of Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

### STEP 6 — DEFINE X AND Y(Feature and Target)

```python
X = df[['YearsExperience']]
Y = df['Salary']
print(X)
print(Y)
```

### STEP 7 — SPLIT DATA

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", Y_train.shape)
print("y_test shape:", Y_test.shape)
```

### STEP 8 — IMPORT MODEL

```python
from sklearn.linear_model import LinearRegression
```

### STEP 9 — CREATE MODEL

```python
model = LinearRegression()
```

### STEP 10 — TRAIN MODEL

```python
model.fit(X_train,Y_train)
```

### STEP 11 — MODEL DETAILS

```python
print("Intercept:", model.intercept_)
print("Coefficient/slope:",model.coef_)
```

### STEP 12 — MAKE PREDICTIONS

```python
Y_pred = model.predict(X_test)
print(Y_pred)
```

### STEP 13 — COMPARE ACTUAL VS PREDICTED

```python
comparison = pd.DataFrame({
    'Actual Salary': Y_test,
    'Predicted Salary' : Y_pred
})
print(comparison)
```

### STEP 14 — VISUALIZE REGRESSION LINE

```python
plt.scatter(X,Y)
plt.plot(X,model.predict(X),color="RED")

plt.title("Linear Regreion Line")
plt.xlabel("Year of Experience")
plt.ylabel("Salary")
plt.grid()
plt.show()
```

### STEP 15 — MODEL EVALUATION

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

mae = mean_absolute_error(Y_test,Y_pred)
mse = mean_squared_error(Y_test,Y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y_test, Y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```
