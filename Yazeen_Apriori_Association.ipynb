```python

import pandas as pd


from mlxtend.preprocessing import TransactionEncoder


from mlxtend.frequent_patterns import apriori


from mlxtend.frequent_patterns import association_rules
```

```python
df_raw = pd.read_csv('Market_Basket_Optimisation.csv', header=None)
```

```python
transactions = []
for i in range(len(df_raw)):
    transaction = [str(item).strip() for item in df_raw.iloc[i].dropna() if str(item).strip() != 'nan']
    transactions.append(transaction)
```

```python
print(f"Total Transactions: {len(transactions)}\n")
print(transactions[:5])
```

```python

te = TransactionEncoder()


te_array = te.fit(transactions).transform(transactions)


df = pd.DataFrame(te_array, columns=te.columns_)
```

```python
print(df.head())
```

```python
frequent_itemsets = apriori(
    df,
    min_support=0.03,
    use_colnames=True
)
```

```python
# Display the frequent itemsets sorted by support
print(frequent_itemsets.sort_values(by='support', ascending=False))
```

```python
# Generate association rules with minimum confidence threshold of 20% (0.2)
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.2
)
```

```python
# Display all generated rules with full metrics
print(rules)
```

```python
# Filter and display primary columns: antecedents, consequents, support, confidence, lift
print(
    rules[
        [
            'antecedents',
            'consequents',
            'support',
            'confidence',
            'lift'
        ]
    ]
)
```

```python
# Filter rules with strong positive correlation (Lift > 1.2)
strong_rules = rules[
    rules['lift'] > 1.2
]

print(
   strong_rules[
        [
            'antecedents',
            'consequents',
            'support',
            'confidence',
            'lift'
        ]
    ]
)
```

```python
# Create a copy with antecedents, consequents, and lift
result = strong_rules[['antecedents', 'consequents', 'lift']].copy()

# Convert frozensets to clean readable strings
result['Rule'] = (
    result['antecedents'].apply(lambda x: ', '.join(list(x))) +
    " → " +
    result['consequents'].apply(lambda x: ', '.join(list(x)))
)

# Keep relevant columns and rename lift
result = result[['Rule', 'lift']]
result.rename(columns={'lift': 'Lift'}, inplace=True)

# Sort by highest Lift and display
result = result.sort_values(by='Lift', ascending=False)
print(result)
```
